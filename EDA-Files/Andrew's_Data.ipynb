```python
import polars as pl
import matplotlib.pyplot as plt

income_cols = [
    'Under $25,000',
    '$25,000 to $49,999',
    '$50,000 to $74,999',
    '$75,000 to $125,000',
    '$125,000 +'
]

income_labels = [
    'Under $25,000',
    '$25,000 - $49,999',
    '$50,000 - $74,999',
    '$75,000 - $125,000',
    '$125,000+'
]

for col in income_cols:
    if df.schema[col] == pl.String:
        df = df.with_columns(
            pl.col(col)
            .str.replace_all(",", "")
            .cast(pl.Int64)
        )

data = [
    df[col].drop_nulls().to_list()
    for col in income_cols
]

plt.figure(figsize=(10, 6))

plt.boxplot(
    data,
    labels=income_labels
)

medians = [
    df[col].median()
    for col in income_cols
]

for i, median in enumerate(medians, start=1):
    plt.text(
        i + 0.05,
        median,
        f'{median:,.0f}',
        verticalalignment='center',
        fontsize=9
    )

plt.title("Household Income Distribution Across Chicago Community Areas")
plt.xlabel("Income Bracket")
plt.ylabel("Number of Households per Community Area")

plt.yticks(range(0, 13001, 1000))

plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.5
)

plt.xticks(rotation=45)


plt.show()
```

Shape: all brackets are right skewed
Center: The median househould dont get above 2000 but stil vary but hovers between about 800 and 1600
Spread: 125,000 has the gratest variability and lonest whiskers and the under 250000 have smaller spreads. It seems wealthier income communites tend to group more together and have a larger amount of households that live in one negiborhood than less affluent ones

```python
import polars as pl
import matplotlib.pyplot as plt


income_cols = [
    'Under $25,000',
    '$25,000 to $49,999',
    '$50,000 to $74,999',
    '$75,000 to $125,000',
    '$125,000 +'
]

income_labels = [
    'Under $25,000',
    '$25,000 - $49,999',
    '$50,000 - $74,999',
    '$75,000 - $125,000',
    '$125,000+'
]



for col in income_cols:
    if df.schema[col] == pl.String:
        df = df.with_columns(
            pl.col(col)
            .str.replace_all(",", "")
            .cast(pl.Int64)
        )


neighborhood_col = "Community Area"


df = df.with_columns(
    pl.sum_horizontal(income_cols).alias("Total Households")
)


for col in income_cols:
    df = df.with_columns(
        (pl.col(col) / pl.col("Total Households") * 100)
        .alias(col + " %")
    )


df = df.sort("$125,000 + %", descending=True)

neighborhoods = df[neighborhood_col].to_list()

plt.figure(figsize=(14, 8))

bottom = [0] * len(neighborhoods)

for col in income_cols:
    values = df[col + " %"].to_list()

    plt.bar(
        neighborhoods,
        values,
        bottom=bottom,
        label=col
    )

    bottom = [
        bottom[i] + values[i]
        for i in range(len(values))
    ]

plt.title("Household Income Distribution by Chicago Community Area")
plt.xlabel("Community Area")
plt.ylabel("Percentage of Households")

plt.xticks(rotation=90)

plt.legend(title="Income Bracket", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

The chart shows vairation of socioeconimoc across Chicago, with some areas having a overwhelming concetration of high-income households and the other way around with lower income households.
