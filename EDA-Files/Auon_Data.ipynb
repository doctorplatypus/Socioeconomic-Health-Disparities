# Profiling the Raw Data

```python

print("Rows:", df.height)
print("Columns:", df.width)
print("Shape:", df.shape)
df.schema
```

It would appear as though the dataset contains a wide range of data pertaining to emissions and general demographic statistics. Looking at the website, column variables past Sq_mi seem to be purely for recordkeeping, and should be removed when we do our analysis. Several columns have the same value per row/community, and so they should also be removed.

# Cleaning the data

```python

clean_df = df.drop(
    [
        "OBJECTID",
        "CCA_TXT",
        "ID",
        "ID_1",
        "GEO",
        "TYPE",
        "COUNTY",
        "GRC_MEM",
        "EV_STA_L2",
        "EV_STA_DCF",
        "EV_STA_TOT",
        "SOLSMART",
        "Acres",
        "Square_met",
        "Sq_mi",
        "Sq_ft",
        "ft",
        "Shape__Area",
        "Shape__Length",
    ]
)

```

# Profiling the Clean Data

```python
clean_df.glimpse()
```

This means there are 77 rows (entrees/observations), and 48 columns, or attributes each observation has. Each row is a community area.

```python
clean_df.null_count()
```

While there are no null values in the cleaned up dataset, SOLAR_POT is all dashes, so it could be a missing value placeholder.
