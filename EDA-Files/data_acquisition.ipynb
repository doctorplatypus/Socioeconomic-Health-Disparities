Each row represents a daily average of readings from an air quality sensor at a specific location in Chicago. 
The dataset covers dates from September 2025 to October 2026 (updated daily) at multiple locations in and around Chicago.
The columns of most interest are string columns 'startofperiod' and 'sensor_name' and the float64 PO2 and PM 2.5 measurement columns: 'no2conc24hourmean_raw', 'no2conc24hourmean_value', 'pm2_5concmass24hourmean_raw', 'pm2_5concmass24hourmean_value', 'pm2_5concmass24hourmeanu', 'pm2_5concmass24hourmeanu_1', 'pm2_5concnum24hourmean_raw', 'pm2_5concnum24hourmean_value'.

The sample CSV file in the /data folder is only 1,000 rows, a small subset of the full dataset which has been downloaded here from the API endpoint.

Important note: The columns starting with 'bc' are all blank because the Black Carbon sensors used by the city are currently offline.
